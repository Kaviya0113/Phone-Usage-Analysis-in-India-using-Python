- Average Salary and Age by Department:
- Find the Top 3 highest-paid employees
- Apply 10% tax if salary &lt; 60,000, else 20%
- Create visualizations( Univariate, bivariate, Multivariate Analysis)

```python
import pandas as pd
import numpy as np
import datetime as dt
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
df = {'Name': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],'Age': [25, 30, 35, 40, 29, 45, 39, 33],'Department': ['HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR'],'Salary': [50000, 60000, 70000, 80000, 52000, 80500, 79888, 54330],'JoinDate': ['2019-04-01', '2020-05-10', '2018-07-23', '2021-08-15','2019-11-11', '2019-08-13', '2022-08-17', '2023-12-11']}
df = pd.DataFrame(df)
```

```python
df.info()
```

```python
df.to_csv('employee_data.csv', index=False)
print("CSV file 'employee_data.csv' created successfully.\n")
```

```python
it_high_salary = df[(df['Department'] == 'IT') & (df['Salary'] > 60000)]
print("IT department employees with higher salary:\n", it_high_salary, "\n")
```

```python
np.mean(df[['Salary','Age']],axis=0)
```

```python
# Univariate:
import plotly.express as pt
fig = pt.pie(data_frame=df,names='Age')
fig.show()
```

```python
# Univariate:
sns.kdeplot(data=df['Salary'],fill=True)
plt.title('Density Plot')
plt.show()
```

```python
# Bivariate:
plt.figure(figsize=(5,4)) 
sns.barplot(x='Age',y='Salary',hue='Age',data=df,palette=['#FF8999','#98E']) # Palette-->color
plt.title('Age by Salary')
plt.show()
```

```python
# Multivariate:
plt.figure(figsize=(10,15))
sns.pairplot(df)
plt.title('Pair Plot')
plt.show()
```
